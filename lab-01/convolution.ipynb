This notebook implements 2D filtering from scratch using NumPy. A custom convolution function applies blur, sharpen, and Sobel kernels to an image using zero-padding and sliding windows. The results are then compared with OpenCV’s `cv2.filter2D` to verify the implementation and observe differences in edge handling and kernel orientation.

```python
import numpy as np
import cv2
```

```python
def convolve2d(image, kernel):
    image = np.asarray(image)
    kernel = np.asarray(kernel)

    if image.ndim != 2:
        raise ValueError(f"Expected a 2D image, but received shape {image.shape}")
    if kernel.ndim != 2:
        raise ValueError(f"Expected a 2D kernel, but received shape {kernel.shape}")

    image_height, image_width = image.shape
    kernel_height, kernel_width = kernel.shape

    pad_height = kernel_height // 2
    pad_width = kernel_width // 2

    padded_image = np.pad(image, ((pad_height, pad_height), (pad_width, pad_width)), mode='constant')

    convolved_image = np.zeros_like(image, dtype=np.float64)

    for i in range(image_height):
        for j in range(image_width):
            region = padded_image[i:i + kernel_height, j:j + kernel_width]
            convolved_value = np.sum(region * kernel)
            convolved_image[i, j] = convolved_value

    return convolved_image
```

```python
image_path = 'image.jpg'
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if image is None:
    rng = np.random.default_rng(42)
    image = rng.integers(0, 256, size=(256, 256), dtype=np.uint8)
    cv2.imwrite(image_path, image)

blur_kernel = np.array([[1, 1, 1],
                        [1, 1, 1],
                        [1, 1, 1]], dtype=np.float64) / 9.0

blurred_image = convolve2d(image, blur_kernel)

opencv_blurred_image = cv2.filter2D(image, -1, blur_kernel)

cv2.imwrite('blurred_image.jpg', np.clip(blurred_image, 0, 255).astype(np.uint8))
cv2.imwrite('opencv_blurred_image.jpg', opencv_blurred_image)

print('Loaded image shape:', image.shape)
print('Custom blur saved to blurred_image.jpg')
print('OpenCV blur saved to opencv_blurred_image.jpg')
```
